# 260814 미션 🟡 — 지시문(system) 실험

같은 질문에 **지시만** 바꿔 본다. `내지시문들` 과 `question` 을 마음대로 바꾸는 것이 미션이다.
제출: 가장 재미있던 비교 1개 캡처 + 한 줄.

```python
# 실험 A — 지시문 세 가지로 같은 질문
from openai import OpenAI

client = OpenAI()
API_MODEL = "gpt-5.4-nano"

question = "오늘 볼 만한 영화를 추천해 줘."

내지시문들 = [
    "너는 마블 영화의 모든 이스터에그를 기억하는 MCU 전문가야.",
    "너는 결말을 예측하는 것을 즐기는 추리 소설 작가야.",
    "너는 웃음이 끊이지 않는 코미디 영화를 가장 좋아하는 영화 애호가야.",
]

for sys_msg in 내지시문들:
    r = client.chat.completions.create(
        model=API_MODEL,
        messages=[
            {"role": "system", "content": sys_msg},
            {"role": "user", "content": question},
        ],
        max_completion_tokens=400,
    )
    print(f"\n[지시] {sys_msg}")
    print(r.choices[0].message.content)
```

```python
# 실험 B — 가짜 과거를 심으면 모델이 따라 한다 (few-shot 맛보기)
# 오늘 배운 그 배열이다. 예시 문답을 미리 심어 두면, 모델은 그 말투를 이어 간다.
예시문답 = [
    {"role": "user", "content": "마블"},
    {"role": "assistant", "content": "마블은 상상력이 폭발하는 거대한 우주."},

    {"role": "user", "content": "추리"},
    {"role": "assistant", "content": "추리는 흩어진 단서를 연결하는 퍼즐 게임."},
]

내단어 = "로맨틱 코미디"

r = client.chat.completions.create(
    model=API_MODEL,
    messages=예시문답 + [{"role": "user", "content": 내단어}],
    max_completion_tokens=200,
)
print(f"{내단어} → {r.choices[0].message.content}")
# 예시를 0개로 줄이면? 예시의 말투를 바꾸면? — 배열에 심은 과거가 답의 성격을 만든다.
```
